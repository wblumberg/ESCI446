# How to Filter Pandas DataFrames with `query()`:

This guide shows multiple examples of how to use the Pandas `query()` function to filter meteorological data.  The `query()` function provides a clean, readable way to filter DataFrames using string expressions.

## Sample Dataset Setup

First, let's create a sample weather dataset:

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create sample meteorological data
data = {
    'date': pd.date_range('2025-01-01', periods=100, freq='H'),
    'temperature': np.random.uniform(-10, 35, 100),  # °C
    'humidity': np.random.uniform(20, 100, 100),     # %
    'pressure': np.random. uniform(980, 1030, 100),   # hPa
    'wind_speed': np.random.uniform(0, 25, 100),     # m/s
    'precipitation': np.random.uniform(0, 10, 100),  # mm
    'station':  np.random.choice(['KMSP', 'KORD', 'KJFK', 'KSEA'], 100)
}

df = pd.DataFrame(data)

# Display the first few rows
print("Dataset shape:", df.shape)
df.head(10)
```

## Basic Filtering Examples

### 1. Simple Comparison
Filter for temperatures above freezing:

```python
# Find all records where temperature is above 0°C
warm_weather = df.query('temperature > 0')

print(f"Total records: {len(df)}")
print(f"Warm weather records: {len(warm_weather)}")
warm_weather.head()
```

### 2. Multiple Conditions (AND)
Filter for ideal flying conditions:

```python
# Low wind and good visibility conditions
good_flying = df.query('wind_speed < 10 and humidity < 70')

print(f"Good flying conditions: {len(good_flying)} records")
good_flying.head()
```

### 3. Multiple Conditions (OR)
Filter for extreme conditions:

```python
# Very hot or very cold conditions
extreme_temps = df.query('temperature > 30 or temperature < -5')

print(f"Extreme temperature records: {len(extreme_temps)}")
extreme_temps.head()
```

### 4. Range Filtering
Filter for comfortable weather conditions:

```python
# Temperature between 18-25°C and low wind
comfortable = df.query('18 <= temperature <= 25 and wind_speed < 5')

print(f"Comfortable conditions: {len(comfortable)} records")
comfortable.head()
```

## Intermediate Examples

### 5. String Matching
Filter by weather station:

```python
# Get data from Chicago O'Hare
chicago_data = df.query("station == 'KORD'")

print(f"Chicago O'Hare records: {len(chicago_data)}")
chicago_data.head()
```

```python
# Multiple stations
major_airports = df.query("station in ['KJFK', 'KORD']")

print(f"Major airports records: {len(major_airports)}")
print("\nStation distribution:")
print(major_airports['station'].value_counts())
```

### 6. Precipitation Events
Identify different precipitation intensities:

```python
# Light rain (0. 1-2. 5 mm/hr)
light_rain = df.query('0.1 <= precipitation < 2.5')

# Heavy rain (>7. 6 mm/hr)
heavy_rain = df.query('precipitation >= 7.6')

# No precipitation
dry_conditions = df.query('precipitation == 0')

print(f"Light rain events: {len(light_rain)}")
print(f"Heavy rain events: {len(heavy_rain)}")
print(f"Dry conditions: {len(dry_conditions)}")
```

### 7. High Pressure Systems
Filter for anticyclonic conditions:

```python
# High pressure with stable conditions
high_pressure = df.query('pressure > 1013 and wind_speed < 8')

print(f"High pressure systems: {len(high_pressure)} records")
print(f"\nAverage conditions during high pressure:")
print(high_pressure[['temperature', 'humidity', 'pressure', 'wind_speed']]. mean())
```

### 8. Storm Conditions
Identify potential severe weather:

```python
# Low pressure with high winds
storm_potential = df.query('pressure < 990 and wind_speed > 15')

print(f"Storm potential records: {len(storm_potential)}")
storm_potential.head()
```

## Advanced Examples

### 9. Using Variables
Pass Python variables into queries using `@`:

```python
# Define thresholds
temp_threshold = 20
wind_threshold = 12

# Use variables in query
filtered = df.query('temperature > @temp_threshold and wind_speed < @wind_threshold')

print(f"Filtered records:  {len(filtered)}")
print(f"Criteria:  temperature > {temp_threshold}°C and wind_speed < {wind_threshold} m/s")
filtered.head()
```

### 10. Heat Index Calculations
Combine conditions to identify uncomfortable conditions:

```python
# Hot and humid (feels-like temperature warning)
uncomfortable = df.query('temperature > 27 and humidity > 70')

print(f"Uncomfortable conditions: {len(uncomfortable)} records")
uncomfortable.head()
```

### 11. Inversions and Stable Conditions
Identify temperature inversion potential:

```python
# Cold, calm conditions favorable for inversions
inversion_conditions = df.query('temperature < 5 and wind_speed < 3 and pressure > 1020')

print(f"Inversion potential:  {len(inversion_conditions)} records")
inversion_conditions.head()
```

### 12. Negation (NOT)
Exclude certain conditions:

```python
# Everything except Seattle station
not_seattle = df.query("station != 'KSEA'")

print(f"Non-Seattle records: {len(not_seattle)}")
print("\nStation distribution:")
print(not_seattle['station'].value_counts())
```

```python
# Not in a list
coastal_stations = df.query("station not in ['KSEA', 'KJFK']")

print(f"Non-coastal station records: {len(coastal_stations)}")
print("\nStation distribution: ")
print(coastal_stations['station'].value_counts())
```

### 13. Complex Boolean Logic
Combine multiple conditions with parentheses:

```python
# Either foggy conditions OR storm conditions
hazardous = df.query('(humidity > 95 and wind_speed < 2) or (wind_speed > 20 and precipitation > 5)')

print(f"Hazardous conditions: {len(hazardous)} records")
hazardous.head()
```

### 14. Date/Time Filtering
Filter by specific date criteria:

```python
# First, ensure date is datetime type
df['date'] = pd.to_datetime(df['date'])

# Extract hour and day components
df['hour'] = df['date'].dt.hour
df['day'] = df['date'].dt.day

# Filter by specific date criteria
morning_data = df.query('hour >= 6 and hour <= 10')

print(f"Morning data (6-10 AM): {len(morning_data)} records")
morning_data.head()
```

```python
# Mid-month data
mid_month = df.query('day >= 10 and day <= 20')

print(f"Mid-month data:  {len(mid_month)} records")
mid_month.head()
```

## Performance Tips

### 15. Column Name Spaces
If your column names have spaces, use backticks to specify the column name:

```python
# Create a copy with column name containing spaces
df_with_spaces = df.copy()
df_with_spaces['dew point'] = np.random.uniform(-5, 20, len(df))

# Query using backticks for column names with spaces
result = df_with_spaces.query('`dew point` > 15')

print(f"Records with dew point > 15°C: {len(result)}")
result[['temperature', 'humidity', 'dew point']].head()
```

### 16. Method Chaining
Combine query with other Pandas operations:

```python
# Filter and then calculate statistics
result = (df.query('temperature > 20 and station == "KORD"')
            .groupby('station')['humidity']
            .agg(['mean', 'min', 'max', 'count']))

print("Statistics for warm days at Chicago O'Hare:")
result
```

## Common Meteorological Use Cases

### 17. Freezing Level Identification

```python
# Near-freezing conditions (important for precipitation type)
freezing_level = df.query('-2 <= temperature <= 2')

print(f"Near-freezing conditions: {len(freezing_level)} records")
freezing_level.head()
```

### 18. Wind Chill Conditions

```python
# Cold and windy (wind chill warning criteria)
wind_chill_warning = df.query('temperature < -10 and wind_speed > 10')

print(f"Wind chill warning conditions: {len(wind_chill_warning)} records")
wind_chill_warning.head()
```

### 19. Drought Monitoring

```python
# Extended dry periods with low humidity
drought_conditions = df.query('precipitation < 0.1 and humidity < 40')

print(f"Drought conditions: {len(drought_conditions)} records")
drought_conditions.head()
```

### 20. Perfect Storm Setup

```python
# Conditions favorable for severe weather development
severe_setup = df.query('pressure < 1000 and temperature > 25 and humidity > 60 and wind_speed > 12')

print(f"Severe weather setup: {len(severe_setup)} records")
severe_setup.head()
```

## Comparison:  query() vs Boolean Indexing

### Traditional Boolean Indexing

```python
# Traditional way
result_traditional = df[(df['temperature'] > 20) & (df['humidity'] < 60)]

print(f"Records using traditional indexing: {len(result_traditional)}")
result_traditional.head()
```

### Using query() (cleaner and more readable)

```python
# Using query()
result_query = df.query('temperature > 20 and humidity < 60')

print(f"Records using query(): {len(result_query)}")
result_query.head()
```

```python
# Verify both methods give the same result
print(f"Both methods return same results: {result_traditional.equals(result_query)}")
```

## Key Advantages of query()

1. **Readability**: More natural syntax, especially for complex conditions
2. **Performance**: Can be faster for large DataFrames
3. **Cleaner code**: Less parentheses and ampersands
4. **String-based**: Easy to construct queries dynamically

## Common Pitfalls

1. **Column names with special characters**: Use backticks
2. **String values**: Must use quotes inside the query string
3. **Boolean operators**: Use `and`, `or`, `not` (not `&`, `|`, `~`)
4. **Variables**:  Remember to use `@` prefix
