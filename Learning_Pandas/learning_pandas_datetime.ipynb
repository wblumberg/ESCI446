### 1.5 Handling Different Date Formats

Dates can be formatted in many ways.  Pandas can parse most formats automatically:

```python
# Different date formats
date_formats = [
    '2025-01-15',           # ISO format (YYYY-MM-DD)
    '01/15/2025',           # US format (MM/DD/YYYY)
    '15-Jan-2025',          # Day-Month-Year with abbreviated month
    'January 15, 2025',     # Full month name
    '2025-01-15 14:30:00'   # With time
]

# Loop over all of the items in the list and convert each of them to datetime 
for date_str in date_formats:
    converted = pd.to_datetime(date_str)
    print(f"Original: {date_str} -> Converted: {converted}")
```

## Part 2: Extracting Datetime Components

Once you have datetime objects, you can extract specific components like year, month, day, etc.

### 2.1 Create Sample Weather Dataset

Let's create a more realistic weather dataset for the examples:

```python
# Create hourly weather data for one week
dates = pd.date_range(start='2025-01-01', periods=168, freq='H')  # 168 hours = 7 days

# Generate realistic temperature data with daily cycle
# Temperature varies by time of day (warmer in afternoon, cooler at night)
hours = np.arange(168)
base_temp = 5  # Base temperature in Celsius
daily_variation = 6 * np.sin((hours % 24 - 6) * 2 * np.pi / 24)  # Daily cycle
random_noise = np.random.normal(0, 1, 168)  # Random variations
temperature = base_temp + daily_variation + random_noise

# Create the DataFrame
weather_df = pd.DataFrame({
    'datetime': dates,
    'temperature': temperature,
    'humidity': np.random.uniform(60, 90, 168),
    'pressure': 1013 + np.random.normal(0, 5, 168),
    'wind_speed': np.abs(np.random.normal(8, 3, 168))
})

# Set datetime as the index (very useful for time series analysis)
weather_df.set_index('datetime', inplace=True)

print("Weather Dataset Created:")
print(f"Shape: {weather_df. shape}")
print(f"Date range: {weather_df.index.min()} to {weather_df.index.max()}")
print("\nFirst 10 rows:")
print(weather_df. head(10).round(2))
```

### 2.2 Extracting Date Components

Let's use various accessors to access different date components from our datetime objects!

```python
# Extract various datetime components
# Note: We use weather_df.index because datetime is our index

weather_df['year'] = weather_df.index.year           # Extract year
weather_df['month'] = weather_df.index.month         # Extract month (1-12)
weather_df['day'] = weather_df.index.day             # Extract day of month (1-31)
weather_df['hour'] = weather_df.index.hour           # Extract hour (0-23)
weather_df['day_of_week'] = weather_df.index.dayofweek  # Day of week (0=Monday, 6=Sunday)
weather_df['day_name'] = weather_df.index.day_name()    # Day name (Monday, Tuesday, etc.)

print("DataFrame with extracted date components:")
print(weather_df.head(10))
```

```python
# If datetime is a column (not index), use .dt accessor
# Let's create an example with datetime as a column
df_example = weather_df.reset_index()  # Move index back to a column

# Now extract components using .dt accessor
df_example['hour_of_day'] = df_example['datetime'].dt.hour
df_example['is_weekend'] = df_example['datetime'].dt.dayofweek >= 5  # Saturday or Sunday

print("\nUsing .dt accessor on a datetime column:")
print(df_example[['datetime', 'hour_of_day', 'is_weekend']].head(10))
```

### 2.3 Practical Example: Average Temperature by Hour of Day

Let's use datetime components to analyze daily temperature patterns:

```python
# Calculate average temperature for each hour of the day
# This shows the diurnal (daily) temperature cycle
hourly_avg_temp = weather_df.groupby('hour')['temperature'].mean()

print("Average Temperature by Hour of Day:")
print("="*50)
for hour, temp in hourly_avg_temp. items():
    # Format hour as HH:00 (e.g., 00:00, 01:00, etc.)
    print(f"{hour: 02d}:00 - {temp:.2f}°C")

# Find warmest and coldest hours
warmest_hour = hourly_avg_temp.idxmax()
coldest_hour = hourly_avg_temp.idxmin()
print(f"\nWarmest hour: {warmest_hour: 02d}:00 ({hourly_avg_temp. max():.2f}°C)")
print(f"Coldest hour:  {coldest_hour:02d}:00 ({hourly_avg_temp.min():.2f}°C)")
```

## Part 3: Date Arithmetic and Calculations

You can perform arithmetic operations with dates (add/subtract time periods).

### 3.1 Adding and Subtracting Time Periods

```python
# Start with a specific date
start_date = pd.Timestamp('2025-01-15 12:00:00')
print(f"Start date: {start_date}")

# Add different time periods using pd. Timedelta
one_day_later = start_date + pd. Timedelta(days=1)
one_week_later = start_date + pd.Timedelta(weeks=1)
three_hours_later = start_date + pd.Timedelta(hours=3)
thirty_minutes_later = start_date + pd.Timedelta(minutes=30)

print(f"\nOne day later: {one_day_later}")
print(f"One week later: {one_week_later}")
print(f"Three hours later: {three_hours_later}")
print(f"Thirty minutes later: {thirty_minutes_later}")
```

```python
# Subtract time periods
one_day_earlier = start_date - pd.Timedelta(days=1)
six_hours_earlier = start_date - pd.Timedelta(hours=6)

print(f"One day earlier: {one_day_earlier}")
print(f"Six hours earlier: {six_hours_earlier}")
```

### 3.2 Calculating Time Differences

Find the duration between two dates:

```python
# Define two dates
observation_time = pd.Timestamp('2025-01-15 14:30:00')
forecast_time = pd.Timestamp('2025-01-16 02:30:00')

# Calculate the difference
time_diff = forecast_time - observation_time

print(f"Observation time: {observation_time}")
print(f"Forecast time: {forecast_time}")
print(f"\nTime difference: {time_diff}")
print(f"Difference in hours: {time_diff.total_seconds() / 3600}")
print(f"Difference in minutes: {time_diff.total_seconds() / 60}")
```

### 3.3 Calculating Lead Time for Forecasts

A practical meteorological example:

```python
# Create forecast dataset
# Forecast is issued at 00Z and valid for different lead times
forecast_issue_time = pd.Timestamp('2025-01-15 00:00:00')

# Create forecast valid times (every 6 hours for 3 days)
lead_times = [6, 12, 18, 24, 30, 36, 42, 48, 54, 60, 66, 72]  # hours
forecast_data = []

for lead_time in lead_times:
    # Calculate valid time by adding lead time to issue time
    valid_time = forecast_issue_time + pd.Timedelta(hours=lead_time)
    
    forecast_data.append({
        'lead_time_hours': lead_time,
        'valid_time': valid_time,
        'forecast_temp': np.random.uniform(0, 10)  # Random forecast temperature
    })

forecast_df = pd.DataFrame(forecast_data)

print(f"Forecast issued:  {forecast_issue_time}")
print("\nForecast lead times:")
print(forecast_df)
```

## Part 4: Filtering Data by Date and Time

Selecting data within specific time periods is crucial for meteorological analysis.

### 4.1 Create a Larger Dataset

Let's create a full year of daily weather data:

```python
# Create one year of daily weather observations
dates = pd.date_range(start='2025-01-01', end='2025-12-31', freq='D')

# Generate seasonal temperature pattern
# Temperature varies sinusoidally throughout the year
day_of_year = np.arange(len(dates))
seasonal_temp = 12 + 10 * np.sin((day_of_year - 80) * 2 * np.pi / 365)
temperature = seasonal_temp + np.random.normal(0, 3, len(dates))

# Create DataFrame
yearly_weather = pd.DataFrame({
    'date': dates,
    'temperature': temperature,
    'precipitation': np.random.exponential(2, len(dates)),
    'wind_speed': np.abs(np.random.normal(8, 3, len(dates)))
})

# Set date as index for easier time-based filtering
yearly_weather.set_index('date', inplace=True)

print(f"Dataset created:  {len(yearly_weather)} days")
print(f"Date range: {yearly_weather.index.min()} to {yearly_weather.index.max()}")
print("\nFirst few rows:")
print(yearly_weather.head())
```

## Part 5: Resampling and Time-Based Aggregation

Resampling converts time series data from one frequency to another (e.g., hourly to daily).

### 5.1 Downsampling (High to Low Frequency)

Convert hourly data to daily averages:

```python
# Our weather_df has hourly data
# Let's convert it to daily data by taking the mean

# Resample to daily ('D') and calculate mean
daily_avg = weather_df[['temperature', 'humidity', 'pressure', 'wind_speed']].resample('D').mean()

print("Daily averages from hourly data:")
print(daily_avg.round(2))
```

```python
# You can use different aggregation functions for different variables
# For example:  daily max temperature and total precipitation
daily_stats = weather_df. resample('D').agg({
    'temperature': ['mean', 'min', 'max'],  # Mean, min, max temperature
    'humidity': 'mean',                      # Average humidity
    'wind_speed': 'max'                      # Maximum wind gust
})

print("\nDaily statistics with multiple aggregations:")
print(daily_stats.round(2))
```

### 5.2 Monthly and Seasonal Aggregations

```python
# Resample yearly data to monthly averages
# 'M' means month-end frequency
monthly_avg = yearly_weather.resample('M').mean()

# Add month name for readability
monthly_avg['month_name'] = monthly_avg.index.month_name()

print("Monthly average conditions for 2025:")
print(monthly_avg.round(2))
```

```python
# Resample to quarterly (seasonal) data
# 'Q' means quarter-end frequency
quarterly_avg = yearly_weather.resample('Q').mean()

# Add quarter labels
quarters = ['Q1 (Winter/Spring)', 'Q2 (Spring/Summer)', 'Q3 (Summer/Fall)', 'Q4 (Fall/Winter)']
quarterly_avg['quarter'] = quarters

print("\nQuarterly (seasonal) average conditions: ")
print(quarterly_avg.round(2))
```

### 5.3 Custom Resampling Periods

```python
# Resample to 3-hour intervals
# This is common for weather forecasts
three_hourly = weather_df[['temperature', 'wind_speed']].resample('3H').mean()

print("3-hourly averages:")
print(three_hourly. round(2).head(16))  # Show first 2 days
```

```python
# Weekly summaries
# 'W' means weekly frequency (week ending on Sunday by default)
weekly_summary = yearly_weather.resample('W').agg({
    'temperature': ['mean', 'min', 'max'],
    'precipitation': 'sum',  # Total weekly precipitation
    'wind_speed': 'mean'
})

print("\nWeekly summaries (first 10 weeks):")
print(weekly_summary.round(2).head(10))
```

### 5.4 Upsampling (Low to High Frequency)

Converting from lower to higher frequency requires filling in missing values:

```python
# Start with daily data and upsample to hourly
# Take first 5 days from yearly_weather
daily_sample = yearly_weather.head(5)[['temperature']]

print("Original daily data:")
print(daily_sample)

# Upsample to hourly and forward-fill the values
# ffill() = forward fill (use the last known value)
hourly_upsampled = daily_sample.resample('H').ffill()

print("\nUpsampled to hourly (forward filled):")
print(hourly_upsampled.head(30))  # Show first 30 hours
```

```python
# Upsample with interpolation (creates smooth transitions)
# interpolate() fills values by estimating between known points
hourly_interpolated = daily_sample.resample('H').interpolate(method='linear')

print("Upsampled to hourly (interpolated):")
print(hourly_interpolated.head(30))
```

## Part 7: Advanced Meteorological Applications

### 7.1 Calculating Growing Degree Days (GDD)

GDD is used in agriculture to predict plant development:

```python
# Growing Degree Days formula: GDD = max(0, (Tmax + Tmin)/2 - Tbase)
# Tbase is typically 10°C for many crops

# Create sample data with daily max and min temperatures
spring_dates = pd.date_range('2025-03-01', '2025-05-31', freq='D')
n_days = len(spring_dates)

# Generate realistic spring temperatures (warming trend)
days_since_start = np.arange(n_days)
base_temp = 8 + 0.15 * days_since_start  # Gradual warming

spring_weather = pd.DataFrame({
    'date': spring_dates,
    'temp_max': base_temp + np.random.uniform(3, 8, n_days),
    'temp_min': base_temp - np.random.uniform(3, 8, n_days)
})

spring_weather. set_index('date', inplace=True)

# Calculate GDD with base temperature of 10°C
base_temperature = 10
spring_weather['avg_temp'] = (spring_weather['temp_max'] + spring_weather['temp_min']) / 2
spring_weather['gdd'] = spring_weather['avg_temp']. apply(lambda x: max(0, x - base_temperature))

# Calculate cumulative GDD
spring_weather['cumulative_gdd'] = spring_weather['gdd'].cumsum()

print("Spring Growing Degree Days:")
print(spring_weather. head(10).round(2))

print(f"\nTotal GDD for spring: {spring_weather['cumulative_gdd'].iloc[-1]:.2f}")
```

### 7.2 Identifying Heat Waves

Detect periods of consecutive hot days:

```python
# Heat wave definition:  3 or more consecutive days with temp > 30°C
heat_threshold = 30
min_consecutive_days = 3

# Create a hot day indicator
yearly_weather['is_hot'] = yearly_weather['temperature'] > heat_threshold

# Find consecutive hot days using groupby
# This creates groups of consecutive True/False values
yearly_weather['hot_day_group'] = (yearly_weather['is_hot'] != yearly_weather['is_hot']. shift()).cumsum()

# Count consecutive days in each group
yearly_weather['consecutive_hot_days'] = yearly_weather. groupby('hot_day_group')['is_hot'].transform('sum')

# Identify heat waves (hot days that are part of 3+ day streaks)
yearly_weather['heat_wave'] = (yearly_weather['is_hot']) & (yearly_weather['consecutive_hot_days'] >= min_consecutive_days)

# Find heat wave periods
heat_wave_days = yearly_weather[yearly_weather['heat_wave']]

print(f"Heat wave analysis (temp > {heat_threshold}°C for {min_consecutive_days}+ days):")
print(f"Total heat wave days: {heat_wave_days.shape[0]}")

if len(heat_wave_days) > 0:
    print("\nHeat wave periods:")
    print(heat_wave_days[['temperature', 'consecutive_hot_days']].round(2))
else:
    print("\nNo heat waves detected in this dataset. ")
```

### 7.3 Calculating Moving Averages

Smooth data to see trends:

```python
# Calculate rolling (moving) averages
# 7-day moving average smooths out daily fluctuations
yearly_weather['temp_7day_avg'] = yearly_weather['temperature'].rolling(window=7, center=True).mean()

# 30-day moving average shows monthly trends
yearly_weather['temp_30day_avg'] = yearly_weather['temperature'].rolling(window=30, center=True).mean()

# Show sample with all three values
print("Temperature with moving averages:")
print(yearly_weather[['temperature', 'temp_7day_avg', 'temp_30day_avg']].iloc[30:40].round(2))
```

### 7.4 Finding Record Temperatures by Day of Year

Compare current observations to historical records:

```python
# Add day of year (1-365)
yearly_weather['day_of_year'] = yearly_weather.index.dayofyear

# Find the record high temperature for each day of the year
record_highs = yearly_weather.groupby('day_of_year')['temperature'].max()
record_lows = yearly_weather.groupby('day_of_year')['temperature'].min()

# Create a summary for first 10 days of the year
record_summary = pd.DataFrame({
    'day_of_year': range(1, 11),
    'record_high':  record_highs.iloc[0:10]. values,
    'record_low': record_lows.iloc[0:10].values
})

print("Record temperatures for first 10 days of year:")
print(record_summary.round(2))
```

### 7.5 Calculating Climate Normals

30-year averages by day of year (climatology):

```python
# Calculate average temperature for each day of year
# This represents the "normal" or expected temperature
climate_normal = yearly_weather.groupby('day_of_year')['temperature'].mean()

# Add climate normal to the dataset
yearly_weather['normal_temp'] = yearly_weather['day_of_year']. map(climate_normal)

# Calculate departure from normal (anomaly)
yearly_weather['temp_anomaly'] = yearly_weather['temperature'] - yearly_weather['normal_temp']

print("Temperature with climate normal and anomaly:")
print(yearly_weather[['temperature', 'normal_temp', 'temp_anomaly']].head(10).round(2))

# Find the day with the largest positive anomaly
warmest_anomaly_idx = yearly_weather['temp_anomaly'].idxmax()
warmest_anomaly = yearly_weather.loc[warmest_anomaly_idx]

print(f"\nLargest warm anomaly: ")
print(f"Date: {warmest_anomaly_idx. strftime('%Y-%m-%d')}")
print(f"Temperature: {warmest_anomaly['temperature']:.2f}°C")
print(f"Normal:  {warmest_anomaly['normal_temp']:.2f}°C")
print(f"Anomaly: +{warmest_anomaly['temp_anomaly']:.2f}°C")
```

### 7.6 Precipitation Accumulation Periods

Calculate running totals over different time periods:

```python
# Calculate cumulative precipitation
yearly_weather['cumulative_precip'] = yearly_weather['precipitation'].cumsum()

# Calculate 7-day precipitation total (rolling sum)
yearly_weather['precip_7day'] = yearly_weather['precipitation'].rolling(window=7).sum()

# Calculate 30-day precipitation total
yearly_weather['precip_30day'] = yearly_weather['precipitation'].rolling(window=30).sum()

# Show sample data
print("Precipitation accumulation analysis:")
print(yearly_weather[['precipitation', 'precip_7day', 'precip_30day', 'cumulative_precip']].iloc[30:40].round(2))

print(f"\nTotal annual precipitation: {yearly_weather['cumulative_precip'].iloc[-1]:.2f} mm")
```

### 7.7 Frost and Freeze Dates

Important for agriculture - first and last freeze dates:

```python
# Identify freezing days (temperature below 0°C)
yearly_weather['is_freeze'] = yearly_weather['temperature'] < 0

# Find all freeze days
freeze_days = yearly_weather[yearly_weather['is_freeze']]

# If the number of days where there is freezing is greater than zero, perform analysis
if len(freeze_days) > 0:
    # First freeze (first day of winter freezing season)
    first_freeze = freeze_days.index[0]
    
    # Last freeze (last day of winter freezing season)
    last_freeze = freeze_days. index[-1]
    
    # Calculate growing season length (days between last spring freeze and first fall freeze)
    growing_season_days = (first_freeze - last_freeze).days
    
    print("Freeze date analysis:")
    print(f"Total freeze days: {len(freeze_days)}")
    print(f"First freeze: {first_freeze.strftime('%Y-%m-%d')}")
    print(f"Last freeze: {last_freeze.strftime('%Y-%m-%d')}")
    
    print("\nFirst 5 freeze days:")
    print(freeze_days[['temperature']].head())
else:
    print("No freezing temperatures in this dataset.")
```
