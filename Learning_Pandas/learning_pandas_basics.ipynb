## Part 2: Reading Data from CSV Files

### The `pd.read_csv()` Function

The primary function for loading data is `pd.read_csv()`. It reads comma-separated values from a file and returns a DataFrame—Pandas' fundamental data structure.

**Common parameters:**
- `filepath_or_buffer`: Path to the CSV file
- `delimiter` or `sep`: Character separating values (default is comma)
- `header`: Row number to use as column names (default=0, the first row)
- `names`: Custom column names
- `index_col`: Column to use as row index
- `skiprows`: Number of rows to skip at the start
- `nrows`: Number of rows to read
- `dtype`: Data types for columns
- `parse_dates`: Columns to parse as datetime

### Example: Loading a Dataset

Let's load a sample dataset. We'll use the tornado dataset available in the datasets folder.

````python
# Load the tornado dataset
df = pd.read_csv('/share/ESCI446/example_datasets/1950-2021_actual_tornadoes.csv')

# If the file has a different delimiter (e.g., tab-separated)
# df = pd.read_csv('path/to/file.csv', delimiter='\t')

print("Dataset loaded successfully!")
print(f"Shape: {df.shape}")
````

## Part 3: Understanding DataFrame Structure

A DataFrame is a 2D table with:
- **Rows**: Observations or records
- **Columns**: Variables or features
- **Index**: Row labels (usually sequential integers)

### 3.1 View the First Few Rows: `head()`

The `.head()` method displays the first n rows (default=5). This gives you a quick sense of what the data looks like.

````python
# View the first 5 rows
df.head()

# View the first 10 rows
# df.head(10)
````

### 3.2 View the Last Few Rows: `tail()`

The `.tail()` method displays the last n rows. Useful for checking recent data or finding patterns near the end of your dataset.

````python
# View the last 5 rows
df.tail()

# View the last 10 rows
# df.tail(10)
````

### 3.3 Get DataFrame Shape: `.shape`

The `.shape` attribute returns a tuple of (number of rows, number of columns). This tells you the size of your dataset.

````python
# Get the number of rows and columns
rows, columns = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {columns}")
````

### 3.4 View Column Names: `.columns`

The `.columns` attribute lists all column names. This is essential for knowing which variables are available in your dataset.

````python
# View all column names
print("Column names:")
print(df.columns.tolist())

# Or just display them directly
df.columns
````

### 3.5 Get Data Type Information: `.dtypes` and `.info()`

Data types are crucial—they determine what operations you can perform on columns.

**Common dtypes:**
- `int64`: Integer numbers
- `float64`: Decimal numbers
- `object`: Strings or mixed types
- `bool`: Boolean (True/False)
- `datetime64`: Dates and times
- `category`: Categorical data (limited categories)

The `.dtypes` attribute shows the type of each column. The `.info()` method provides a more detailed summary.

````python
# View data types for each column
print("Data types:")
print(df.dtypes)
print("\n")

# Get detailed information
print("Detailed DataFrame information:")
df.info()
````

### 3.6 Get Basic Statistics: `.describe()`

The `.describe()` method provides summary statistics for numerical columns:
- `count`: Non-null values
- `mean`: Average
- `std`: Standard deviation
- `min`: Minimum value
- `25%, 50%, 75%`: Quartiles (percentiles)
- `max`: Maximum value

This gives you a quick overview of the distribution and range of your numerical variables.

````python
# Get summary statistics
df.describe()

# Include all columns (including non-numeric)
# df.describe(include='all')
````

## Part 4: Selecting and Accessing Data

There are multiple ways to select data from a DataFrame. Choosing the right method depends on what you're trying to access.

### 4.1 Select a Single Column

You can select a column using bracket notation or dot notation. The result is a Series (1D array-like object).

````python
# Method 1: Using brackets (works with any column name)
year_column = df['yr']
print(type(year_column))
print(year_column.head())

# Method 2: Using dot notation (only works if column name is a valid identifier)
# year_column = df.Year
# print(year_column.head())
````

### 4.2 Select Multiple Columns

Pass a list of column names to get a new DataFrame with only those columns.

````python
# Select multiple columns
subset = df[['yr', 'mo', 'dy']]
print(type(subset))
subset.head()
````

### 4.3 Select Rows by Index: `.iloc[]`

The `.iloc[]` accessor uses integer-location based indexing (like Python lists):
- `df.iloc[row_index]`: Select a single row
- `df.iloc[start:end]`: Select rows by slice (end is exclusive)
- `df.iloc[[i, j, k]]`: Select specific rows by list of indices

````python
# Select the first row (index 0)
print("First row:")
print(df.iloc[0])
print("\n")

# Select rows 0-4 (5 rows total)
print("Rows 0-4:")
print(df.iloc[0:5])
print("\n")

# Select specific rows
print("Rows 0, 2, and 4:")
print(df.iloc[[0, 2, 4]])
````

### 4.4 Select Rows by Label: `.loc[]`

The `.loc[]` accessor uses label-based indexing (like dictionaries):
- `df.loc[row_label]`: Select a single row by its index label
- `df.loc[start:end]`: Select rows by label (end is inclusive, unlike iloc)
- `df.loc[condition]`: Select rows matching a condition

````python
# Assume the index is the default integer index (0, 1, 2, ...)
# Select the row with index label 0
print("Row with index 0:")
print(df.loc[0])
print("\n")

# Select rows with index labels 0-4 (both endpoints included)
print("Rows with index 0-4:")
print(df.loc[0:4])
````

### 4.5 Select Specific Cell: `df.at[]` and `df.iat[]`

To access a specific cell (one value):
- `df.at[row_label, column_name]`: Access by labels
- `df.iat[row_index, column_index]`: Access by integer positions

````python
# Access a specific cell using labels
value = df.at[0, 'yr']
print(f"Year at row 0: {value}")

# Access a specific cell using integer positions
value = df.iat[0, 0]  # First row, first column
print(f"First cell value: {value}")
````

### 4.6 Filter Rows Using Conditions

One of the most powerful features: selecting rows that meet specific conditions. Create a boolean mask and pass it to `.loc[]` or use direct indexing.

````python
# Find all tornadoes from a specific year
year_2020 = df[df['yr'] == 2020]
print(f"Number of tornadoes in 2020: {len(year_2020)}")
print(year_2020.head())
print("\n")

# Find tornadoes with casualties above a threshold
high_casualties = df[df['inj'] > 10]
print(f"Number of tornadoes with > 10 injuries: {len(high_casualties)}")
print(high_casualties.head())
````

### 4.7 Combine Multiple Conditions

Use logical operators (`&` for AND, `|` for OR, `~` for NOT) to combine conditions. Remember to use parentheses around each condition.

````python
# Tornadoes in spring (April-June) with high casualties
spring_months = [4, 5, 6]
high_casualty_spring = df[(df['mo'].isin(spring_months)) & (df['fat'] > 5)]
print(f"High-casualty spring tornadoes: {len(high_casualty_spring)}")
print(high_casualty_spring.head())
print("\n")

# Tornadoes that EITHER occurred before 1960 OR in 2020
old_or_recent = df[(df['yr'] < 1960) | (df['yr'] == 2020)]
print(f"Tornadoes from before 1960 or in 2020: {len(old_or_recent)}")
````

## Part 5: Descriptive Statistics and Summaries

Pandas provides many methods to compute statistics on your data.

### 5.1 Statistics for Entire DataFrame

````python
# Mean of all numerical columns
print("Mean values:")
print(df.mean(numeric_only=True))
print("\n")

# Sum of all numerical columns
print("Sum of values:")
print(df.sum(numeric_only=True))
print("\n")

# Median
print("Median values:")
print(df.median(numeric_only=True))
````

### 5.2 Statistics for Individual Columns

````python
# Common statistics for a single column
deaths_column = df['fat']

print(f"Mean deaths per tornado: {deaths_column.mean():.2f}")
print(f"Median deaths: {deaths_column.median():.2f}")
print(f"Standard deviation: {deaths_column.std():.2f}")
print(f"Minimum deaths: {deaths_column.min()}")
print(f"Maximum deaths: {deaths_column.max()}")
print(f"Total deaths: {deaths_column.sum()}")
````

### 5.3 Counting and Unique Values

````python
# Count non-null values
print("Count of non-null values:")
print(df.count())
print("\n")

# Count unique values
print("Number of unique values per column:")
print(df.nunique())
print("\n")

# Get unique values for a column
print(f"Unique years: {sorted(df['yr'].unique())}")
````

### 5.4 Value Counts: Distribution of Categorical Data

The `.value_counts()` method shows the frequency of each unique value, useful for understanding categorical variables.

````python
# Count how many tornadoes per month
print("Tornadoes by month:")
print(df['mo'].value_counts().sort_index())
print("\n")

# Most common month (top 5)
print("Top 5 months with most tornadoes:")
print(df['mo'].value_counts().head(5))
````

## Part 6: Handling Missing Data

Real-world datasets often have missing values. It's important to detect and handle them appropriately.

### 6.1 Detect Missing Values

````python
# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())
print("\n")

# Percentage of missing values
print("Percentage of missing values:")
print((df.isnull().sum() / len(df) * 100).round(2))
print("\n")

# Total number of missing values
print(f"Total missing values: {df.isnull().sum().sum()}")
````

### 6.2 Handle Missing Values

Common strategies:
- **Remove**: Delete rows or columns with missing values
- **Fill**: Replace with a default value (mean, median, forward fill, etc.)
- **Impute**: Use more sophisticated methods

````python
# Remove rows with any missing values
df_no_missing = df.dropna()
print(f"Rows after removing any NaN: {len(df_no_missing)}")
print("\n")

# Remove rows where a specific column is missing
df_deaths_present = df.dropna(subset=['fat'])
print(f"Rows with Deaths data: {len(df_deaths_present)}")
print("\n")

# Fill missing values with a constant
df_filled = df.fillna(0)
print(f"After filling NaN with 0: {df_filled.isnull().sum().sum()} missing values")
````

## Part 7: Practical Investigation Workflow

Here's a typical workflow for investigating a new dataset:

1. Load the data
2. Check shape and basic info
3. View sample rows
4. Examine data types and missing values
5. Compute summary statistics
6. Explore relationships with filtering and grouping

Let's apply this to our tornado dataset:

````python
# Step 1: Overview
print("=== DATASET OVERVIEW ===")
print(f"Shape: {df.shape}")
print(f"\nFirst few rows:")
print(df.head())
````

````python
# Step 2: Data types and info
print("=== DATA TYPES AND INFO ===")
df.info()
````

````python
# Step 3: Missing data
print("=== MISSING VALUES ===")
missing_summary = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': df.isnull().sum(),
    'Missing_Percent': (df.isnull().sum() / len(df) * 100).round(2)
})
print(missing_summary[missing_summary['Missing_Count'] > 0])
````

````python
# Step 4: Summary statistics
print("=== SUMMARY STATISTICS ===")
print(df.describe())
````

````python
# Step 5: Explore key questions
print("=== EXPLORATORY QUESTIONS ===")
print(f"Time range: {df['yr'].min()} to {df['yr'].max()}")
print(f"\nMost deadly tornado: {df['fat'].max()} deaths")
print(f"Total deaths in dataset: {df['fat'].sum()}")
print(f"\nAverage deaths per tornado: {df['fat'].mean():.2f}")
print(f"\nTornadoes by decade:")
df['Decade'] = (df['yr'] // 10 * 10).astype(str) + 's'
print(df['Decade'].value_counts().sort_index())
````
